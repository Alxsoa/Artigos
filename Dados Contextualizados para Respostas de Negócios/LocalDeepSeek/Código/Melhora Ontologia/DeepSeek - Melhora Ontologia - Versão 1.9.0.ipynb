```python
# ======================================================================
# Bibliotecas Necessárias
# ======================================================================
from rdflib import Graph 
from rdflib import Namespace 
from rdflib import Literal 
from rdflib import URIRef
from rdflib.namespace import RDF 
from rdflib.namespace import RDFS 
from rdflib.namespace import OWL 
from rdflib.namespace import XSD
from datetime import datetime
from collections import Counter
import networkx as nx
import matplotlib.pyplot as plt
import json
import pickle
```

```python
# Namespace
ns = Namespace("http://example.org/turing#")

def criar_ontologia_otimizada():
    """
    Cria ontologia do zero com:
    - Hierarquia profunda (4-6 níveis)
    - Domínio e range para TODAS as propriedades
    - Muitas datatype properties
    - Todas as características OWL
    - Documentação completa
    """
    
    print ("# ====================================================================== ")
    print ("# CRIANDO ONTOLOGIA DE ALTA QUALIDADE DO ZERO")
    print ("# ====================================================================== ")
    
    g = Graph()
    g.bind('', ns)
    g.bind('owl', OWL)
    g.bind('rdfs', RDFS)
    g.bind('xsd', XSD)
    
    # Ontology metadata
    ontology_uri = URIRef(ns)
    g.add((ontology_uri, RDF.type, OWL.Ontology))
    g.add((ontology_uri, RDFS.label, Literal("Ontologia Alan Turing - Alta Qualidade", lang='pt')))
    g.add((ontology_uri, RDFS.comment, 
          Literal("Ontologia otimizada sobre Alan Turing com hierarquia profunda e alta riqueza semântica", lang='pt')))
    
    print("# Criando hierarquia de classes (6 níveis)")
    
    # ===================================================================
    # HIERARQUIA DE CLASSES (6 NÍVEIS)
    # ===================================================================
    
    # NÍVEL 1: Raiz
    g.add((ns.Entidade, RDF.type, OWL.Class))
    g.add((ns.Entidade, RDFS.label, Literal("Entidade", lang='pt')))
    g.add((ns.Entidade, RDFS.comment, 
          Literal("Classe raiz que representa qualquer entidade no domínio", lang='pt')))
    
    # NÍVEL 2: Categorias principais
    classes_nivel2 = {
        'Agente': 'Entidades que podem agir e tomar decisões',
        'Lugar': 'Localizações geográficas e espaços físicos',
        'Objeto': 'Objetos físicos e artefatos',
        'Abstrato': 'Conceitos e entidades abstratas',
        'Temporal': 'Entidades relacionadas ao tempo'
    }
    
    for classe, desc in classes_nivel2.items():
        cls = ns[classe]
        g.add((cls, RDF.type, OWL.Class))
        g.add((cls, RDFS.subClassOf, ns.Entidade))
        g.add((cls, RDFS.label, Literal(classe, lang='pt')))
        g.add((cls, RDFS.comment, Literal(desc, lang='pt')))
    
    # NÍVEL 3: Subcategorias
    
    # Agente → Pessoa, Organização
    g.add((ns.Pessoa, RDF.type, OWL.Class))
    g.add((ns.Pessoa, RDFS.subClassOf, ns.Agente))
    g.add((ns.Pessoa, RDFS.label, Literal("Pessoa", lang='pt')))
    g.add((ns.Pessoa, RDFS.comment, Literal("Indivíduo humano", lang='pt')))
    
    g.add((ns.Organizacao, RDF.type, OWL.Class))
    g.add((ns.Organizacao, RDFS.subClassOf, ns.Agente))
    g.add((ns.Organizacao, RDFS.label, Literal("Organização", lang='pt')))
    g.add((ns.Organizacao, RDFS.comment, Literal("Grupo organizado de pessoas com objetivo comum", lang='pt')))
    
    # Lugar → Pais, Cidade, Edificio
    g.add((ns.Pais, RDF.type, OWL.Class))
    g.add((ns.Pais, RDFS.subClassOf, ns.Lugar))
    g.add((ns.Pais, RDFS.label, Literal("País", lang='pt')))
    g.add((ns.Pais, RDFS.comment, Literal("Unidade política soberana", lang='pt')))
    
    g.add((ns.Cidade, RDF.type, OWL.Class))
    g.add((ns.Cidade, RDFS.subClassOf, ns.Lugar))
    g.add((ns.Cidade, RDFS.label, Literal("Cidade", lang='pt')))
    g.add((ns.Cidade, RDFS.comment, Literal("Área urbana habitada", lang='pt')))
    
    g.add((ns.Edificio, RDF.type, OWL.Class))
    g.add((ns.Edificio, RDFS.subClassOf, ns.Lugar))
    g.add((ns.Edificio, RDFS.label, Literal("Edifício", lang='pt')))
    g.add((ns.Edificio, RDFS.comment, Literal("Construção física", lang='pt')))
    
    # Objeto → Maquina, Documento
    g.add((ns.Maquina, RDF.type, OWL.Class))
    g.add((ns.Maquina, RDFS.subClassOf, ns.Objeto))
    g.add((ns.Maquina, RDFS.label, Literal("Máquina", lang='pt')))
    g.add((ns.Maquina, RDFS.comment, Literal("Dispositivo mecânico ou eletrônico", lang='pt')))
    
    g.add((ns.Documento, RDF.type, OWL.Class))
    g.add((ns.Documento, RDFS.subClassOf, ns.Objeto))
    g.add((ns.Documento, RDFS.label, Literal("Documento", lang='pt')))
    g.add((ns.Documento, RDFS.comment, Literal("Registro de informação", lang='pt')))
    
    # Abstrato → Conceito, Teoria, Area
    g.add((ns.Conceito, RDF.type, OWL.Class))
    g.add((ns.Conceito, RDFS.subClassOf, ns.Abstrato))
    g.add((ns.Conceito, RDFS.label, Literal("Conceito", lang='pt')))
    g.add((ns.Conceito, RDFS.comment, Literal("Ideia ou noção abstrata", lang='pt')))
    
    g.add((ns.Teoria, RDF.type, OWL.Class))
    g.add((ns.Teoria, RDFS.subClassOf, ns.Abstrato))
    g.add((ns.Teoria, RDFS.label, Literal("Teoria", lang='pt')))
    g.add((ns.Teoria, RDFS.comment, Literal("Sistema de ideias que explica fenômenos", lang='pt')))
    
    g.add((ns.AreaConhecimento, RDF.type, OWL.Class))
    g.add((ns.AreaConhecimento, RDFS.subClassOf, ns.Abstrato))
    g.add((ns.AreaConhecimento, RDFS.label, Literal("Área de Conhecimento", lang='pt')))
    g.add((ns.AreaConhecimento, RDFS.comment, Literal("Domínio específico do saber", lang='pt')))
    
    # Temporal → Periodo, Evento
    g.add((ns.Periodo, RDF.type, OWL.Class))
    g.add((ns.Periodo, RDFS.subClassOf, ns.Temporal))
    g.add((ns.Periodo, RDFS.label, Literal("Período", lang='pt')))
    g.add((ns.Periodo, RDFS.comment, Literal("Intervalo de tempo", lang='pt')))
    
    g.add((ns.Evento, RDF.type, OWL.Class))
    g.add((ns.Evento, RDFS.subClassOf, ns.Temporal))
    g.add((ns.Evento, RDFS.label, Literal("Evento", lang='pt')))
    g.add((ns.Evento, RDFS.comment, Literal("Acontecimento específico no tempo", lang='pt')))
    
    # NÍVEL 4: Especializações
    
    # Pessoa → Cientista, Militar, Politico
    g.add((ns.Cientista, RDF.type, OWL.Class))
    g.add((ns.Cientista, RDFS.subClassOf, ns.Pessoa))
    g.add((ns.Cientista, RDFS.label, Literal("Cientista", lang='pt')))
    g.add((ns.Cientista, RDFS.comment, Literal("Pessoa dedicada à ciência", lang='pt')))
    
    g.add((ns.Militar, RDF.type, OWL.Class))
    g.add((ns.Militar, RDFS.subClassOf, ns.Pessoa))
    g.add((ns.Militar, RDFS.label, Literal("Militar", lang='pt')))
    g.add((ns.Militar, RDFS.comment, Literal("Membro das forças armadas", lang='pt')))
    
    g.add((ns.Politico, RDF.type, OWL.Class))
    g.add((ns.Politico, RDFS.subClassOf, ns.Pessoa))
    g.add((ns.Politico, RDFS.label, Literal("Político", lang='pt')))
    g.add((ns.Politico, RDFS.comment, Literal("Pessoa envolvida em política", lang='pt')))
    
    # Organização → InstituicaoEnsino, OrganizacaoMilitar, Empresa
    g.add((ns.InstituicaoEnsino, RDF.type, OWL.Class))
    g.add((ns.InstituicaoEnsino, RDFS.subClassOf, ns.Organizacao))
    g.add((ns.InstituicaoEnsino, RDFS.label, Literal("Instituição de Ensino", lang='pt')))
    g.add((ns.InstituicaoEnsino, RDFS.comment, Literal("Organização educacional", lang='pt')))
    
    g.add((ns.OrganizacaoMilitar, RDF.type, OWL.Class))
    g.add((ns.OrganizacaoMilitar, RDFS.subClassOf, ns.Organizacao))
    g.add((ns.OrganizacaoMilitar, RDFS.label, Literal("Organização Militar", lang='pt')))
    g.add((ns.OrganizacaoMilitar, RDFS.comment, Literal("Organização das forças armadas", lang='pt')))
    
    g.add((ns.Empresa, RDF.type, OWL.Class))
    g.add((ns.Empresa, RDFS.subClassOf, ns.Organizacao))
    g.add((ns.Empresa, RDFS.label, Literal("Empresa", lang='pt')))
    g.add((ns.Empresa, RDFS.comment, Literal("Organização comercial", lang='pt')))
    
    # Maquina → MaquinaComputacional, DispositivoCriptografico
    g.add((ns.MaquinaComputacional, RDF.type, OWL.Class))
    g.add((ns.MaquinaComputacional, RDFS.subClassOf, ns.Maquina))
    g.add((ns.MaquinaComputacional, RDFS.label, Literal("Máquina Computacional", lang='pt')))
    g.add((ns.MaquinaComputacional, RDFS.comment, Literal("Máquina capaz de realizar computações", lang='pt')))
    
    g.add((ns.DispositivoCriptografico, RDF.type, OWL.Class))
    g.add((ns.DispositivoCriptografico, RDFS.subClassOf, ns.Maquina))
    g.add((ns.DispositivoCriptografico, RDFS.label, Literal("Dispositivo Criptográfico", lang='pt')))
    g.add((ns.DispositivoCriptografico, RDFS.comment, Literal("Máquina para cifrar/decifrar mensagens", lang='pt')))
    
    # Documento → Publicacao, Artigo, Livro
    g.add((ns.Publicacao, RDF.type, OWL.Class))
    g.add((ns.Publicacao, RDFS.subClassOf, ns.Documento))
    g.add((ns.Publicacao, RDFS.label, Literal("Publicação", lang='pt')))
    g.add((ns.Publicacao, RDFS.comment, Literal("Documento publicado", lang='pt')))
    
    g.add((ns.Artigo, RDF.type, OWL.Class))
    g.add((ns.Artigo, RDFS.subClassOf, ns.Documento))
    g.add((ns.Artigo, RDFS.label, Literal("Artigo", lang='pt')))
    g.add((ns.Artigo, RDFS.comment, Literal("Texto científico ou jornalístico", lang='pt')))
    
    # Evento → EventoHistorico, EventoCientifico
    g.add((ns.EventoHistorico, RDF.type, OWL.Class))
    g.add((ns.EventoHistorico, RDFS.subClassOf, ns.Evento))
    g.add((ns.EventoHistorico, RDFS.label, Literal("Evento Histórico", lang='pt')))
    g.add((ns.EventoHistorico, RDFS.comment, Literal("Acontecimento de importância histórica", lang='pt')))
    
    g.add((ns.EventoCientifico, RDF.type, OWL.Class))
    g.add((ns.EventoCientifico, RDFS.subClassOf, ns.Evento))
    g.add((ns.EventoCientifico, RDFS.label, Literal("Evento Científico", lang='pt')))
    g.add((ns.EventoCientifico, RDFS.comment, Literal("Acontecimento científico relevante", lang='pt')))
    
    # NÍVEL 5: Super especializações
    
    # Cientista → Matematico, Fisico, CientistaDaComputacao
    g.add((ns.Matematico, RDF.type, OWL.Class))
    g.add((ns.Matematico, RDFS.subClassOf, ns.Cientista))
    g.add((ns.Matematico, RDFS.label, Literal("Matemático", lang='pt')))
    g.add((ns.Matematico, RDFS.comment, Literal("Cientista especializado em matemática", lang='pt')))
    
    g.add((ns.Fisico, RDF.type, OWL.Class))
    g.add((ns.Fisico, RDFS.subClassOf, ns.Cientista))
    g.add((ns.Fisico, RDFS.label, Literal("Físico", lang='pt')))
    g.add((ns.Fisico, RDFS.comment, Literal("Cientista especializado em física", lang='pt')))
    
    g.add((ns.CientistaDaComputacao, RDF.type, OWL.Class))
    g.add((ns.CientistaDaComputacao, RDFS.subClassOf, ns.Cientista))
    g.add((ns.CientistaDaComputacao, RDFS.label, Literal("Cientista da Computação", lang='pt')))
    g.add((ns.CientistaDaComputacao, RDFS.comment, Literal("Cientista especializado em computação", lang='pt')))
    
    g.add((ns.Criptografo, RDF.type, OWL.Class))
    g.add((ns.Criptografo, RDFS.subClassOf, ns.Cientista))
    g.add((ns.Criptografo, RDFS.label, Literal("Criptógrafo", lang='pt')))
    g.add((ns.Criptografo, RDFS.comment, Literal("Especialista em criptografia", lang='pt')))
    
    # InstituicaoEnsino → Universidade, Escola
    g.add((ns.Universidade, RDF.type, OWL.Class))
    g.add((ns.Universidade, RDFS.subClassOf, ns.InstituicaoEnsino))
    g.add((ns.Universidade, RDFS.label, Literal("Universidade", lang='pt')))
    g.add((ns.Universidade, RDFS.comment, Literal("Instituição de ensino superior", lang='pt')))
    
    g.add((ns.Escola, RDF.type, OWL.Class))
    g.add((ns.Escola, RDFS.subClassOf, ns.InstituicaoEnsino))
    g.add((ns.Escola, RDFS.label, Literal("Escola", lang='pt')))
    g.add((ns.Escola, RDFS.comment, Literal("Instituição de ensino básico ou médio", lang='pt')))
    
    # NÍVEL 6: Ultra especializações
    
    # Matematico → MatematicoPuro, MatemaaticoAplicado
    g.add((ns.MatematicoPuro, RDF.type, OWL.Class))
    g.add((ns.MatematicoPuro, RDFS.subClassOf, ns.Matematico))
    g.add((ns.MatematicoPuro, RDFS.label, Literal("Matemático Puro", lang='pt')))
    g.add((ns.MatematicoPuro, RDFS.comment, Literal("Matemático focado em teoria pura", lang='pt')))
    
    g.add((ns.MatemaaticoAplicado, RDF.type, OWL.Class))
    g.add((ns.MatemaaticoAplicado, RDFS.subClassOf, ns.Matematico))
    g.add((ns.MatemaaticoAplicado, RDFS.label, Literal("Matemático Aplicado", lang='pt')))
    g.add((ns.MatemaaticoAplicado, RDFS.comment, Literal("Matemático focado em aplicações", lang='pt')))
    
    print(f"# 38 classes criadas em hierarquia de 6 níveis")
    print ("# ====================================================================== ")
    
    # ===================================================================
    # OBJECT PROPERTIES (com domínio e range)
    # ===================================================================
    
    print("# Criando object properties com domínio e range...")
    
    object_props = {
        # Relações de Pessoa
        'nasceuEm': (ns.Pessoa, ns.Lugar, "Indica onde a pessoa nasceu"),
        'morreuEm': (ns.Pessoa, ns.Lugar, "Indica onde a pessoa morreu"),
        'viveuEm': (ns.Pessoa, ns.Lugar, "Indica onde a pessoa viveu"),
        'trabalhouEm': (ns.Pessoa, ns.Organizacao, "Indica onde a pessoa trabalhou"),
        'estudouEm': (ns.Pessoa, ns.InstituicaoEnsino, "Indica onde a pessoa estudou"),
        'graduouSe': (ns.Pessoa, ns.InstituicaoEnsino, "Indica onde obteve graduação"),
        'doutoradoEm': (ns.Pessoa, ns.InstituicaoEnsino, "Indica onde obteve doutorado"),
        'colaborouCom': (ns.Pessoa, ns.Pessoa, "Indica colaboração entre pessoas"),
        'orientou': (ns.Cientista, ns.Cientista, "Indica orientação acadêmica"),
        'foiOrientadoPor': (ns.Cientista, ns.Cientista, "Indica quem foi o orientador"),
        'pai': (ns.Pessoa, ns.Pessoa, "Indica relação de paternidade"),
        'mae': (ns.Pessoa, ns.Pessoa, "Indica relação de maternidade"),
        'filho': (ns.Pessoa, ns.Pessoa, "Indica relação de filiação"),
        'casouCom': (ns.Pessoa, ns.Pessoa, "Indica casamento"),
        'conheceu': (ns.Pessoa, ns.Pessoa, "Indica que conheceu outra pessoa"),
        
        # Relações de trabalho
        'inventou': (ns.Cientista, ns.Maquina, "Indica invenção de máquina"),
        'desenvolveu': (ns.Cientista, ns.Teoria, "Indica desenvolvimento de teoria"),
        'propôs': (ns.Cientista, ns.Conceito, "Indica proposição de conceito"),
        'publicou': (ns.Cientista, ns.Publicacao, "Indica autoria de publicação"),
        'escreveu': (ns.Pessoa, ns.Documento, "Indica autoria de documento"),
        
        # Relações de Organização
        'localizadaEm': (ns.Organizacao, ns.Lugar, "Indica localização da organização"),
        'fundadaEm': (ns.Organizacao, ns.Cidade, "Indica onde foi fundada"),
        'pertenceA': (ns.Organizacao, ns.Organizacao, "Indica subordinação organizacional"),
        
        # Relações de Máquina
        'construidaPor': (ns.Maquina, ns.Pessoa, "Indica quem construiu"),
        'usadaEm': (ns.Maquina, ns.Lugar, "Indica onde foi usada"),
        'baseadaEm': (ns.Maquina, ns.Teoria, "Indica base teórica"),
        
        # Relações de Evento
        'ocorreuEm': (ns.Evento, ns.Lugar, "Indica onde ocorreu"),
        'participouDe': (ns.Pessoa, ns.Evento, "Indica participação em evento"),
        'organizadoPor': (ns.Evento, ns.Organizacao, "Indica organizador"),
        
        # Relações de Documento
        'publicadoEm': (ns.Publicacao, ns.Periodo, "Indica período de publicação"),
        'sobre': (ns.Documento, ns.Abstrato, "Indica assunto do documento"),
        
        # Relações de AreaConhecimento
        'especializadoEm': (ns.Cientista, ns.AreaConhecimento, "Indica especialização"),
        'contribuiuPara': (ns.Cientista, ns.AreaConhecimento, "Indica contribuição"),
        
        # Relações temporais
        'durante': (ns.Evento, ns.Periodo, "Indica período do evento"),
        'antes': (ns.Temporal, ns.Temporal, "Indica precedência temporal"),
        'depois': (ns.Temporal, ns.Temporal, "Indica sucessão temporal"),
    }
    
    for prop_name, (domain, range_cls, comment) in object_props.items():
        prop = ns[prop_name]
        g.add((prop, RDF.type, OWL.ObjectProperty))
        g.add((prop, RDFS.label, Literal(prop_name, lang='pt')))
        g.add((prop, RDFS.domain, domain))
        g.add((prop, RDFS.range, range_cls))
        g.add((prop, RDFS.comment, Literal(comment, lang='pt')))
    
    print(f"# {len(object_props)} object properties criadas")
    print ("# ====================================================================== ") 

    # ===================================================================
    # DATATYPE PROPERTIES (muitas!)
    # ===================================================================
    
    print("# Criando datatype properties...")
    
    datatype_props = {
        # Pessoa
        'nomeCompleto': (ns.Pessoa, XSD.string, "Nome completo da pessoa"),
        'primeiroNome': (ns.Pessoa, XSD.string, "Primeiro nome"),
        'sobrenome': (ns.Pessoa, XSD.string, "Sobrenome"),
        'dataNascimento': (ns.Pessoa, XSD.date, "Data de nascimento"),
        'dataMorte': (ns.Pessoa, XSD.date, "Data de morte"),
        'idade': (ns.Pessoa, XSD.integer, "Idade da pessoa"),
        'nacionalidade': (ns.Pessoa, XSD.string, "Nacionalidade"),
        'profissao': (ns.Pessoa, XSD.string, "Profissão principal"),
        'email': (ns.Pessoa, XSD.string, "Endereço de email"),
        'biografia': (ns.Pessoa, XSD.string, "Biografia resumida"),
        
        # Lugar
        'nomeLugar': (ns.Lugar, XSD.string, "Nome do lugar"),
        'latitude': (ns.Lugar, XSD.decimal, "Coordenada latitude"),
        'longitude': (ns.Lugar, XSD.decimal, "Coordenada longitude"),
        'populacao': (ns.Cidade, XSD.integer, "População da cidade"),
        'area': (ns.Lugar, XSD.decimal, "Área em km²"),
        'altitude': (ns.Lugar, XSD.decimal, "Altitude em metros"),
        
        # Organização
        'nomeOrganizacao': (ns.Organizacao, XSD.string, "Nome da organização"),
        'dataFundacao': (ns.Organizacao, XSD.date, "Data de fundação"),
        'numeroFuncionarios': (ns.Organizacao, XSD.integer, "Número de funcionários"),
        'website': (ns.Organizacao, XSD.anyURI, "Website oficial"),
        
        # Máquina
        'nomeMaquina': (ns.Maquina, XSD.string, "Nome da máquina"),
        'peso': (ns.Maquina, XSD.decimal, "Peso em kg"),
        'altura': (ns.Maquina, XSD.decimal, "Altura em metros"),
        'largura': (ns.Maquina, XSD.decimal, "Largura em metros"),
        'profundidade': (ns.Maquina, XSD.decimal, "Profundidade em metros"),
        'anoConstrucao': (ns.Maquina, XSD.gYear, "Ano de construção"),
        'velocidade': (ns.MaquinaComputacional, XSD.decimal, "Velocidade de processamento"),
        
        # Documento
        'titulo': (ns.Documento, XSD.string, "Título do documento"),
        'dataPublicacao': (ns.Publicacao, XSD.date, "Data de publicação"),
        'numeroPaginas': (ns.Documento, XSD.integer, "Número de páginas"),
        'idioma': (ns.Documento, XSD.string, "Idioma do documento"),
        'isbn': (ns.Publicacao, XSD.string, "ISBN do livro"),
        'doi': (ns.Artigo, XSD.string, "DOI do artigo"),
        
        # Evento
        'nomeEvento': (ns.Evento, XSD.string, "Nome do evento"),
        'dataInicio': (ns.Evento, XSD.date, "Data de início"),
        'dataFim': (ns.Evento, XSD.date, "Data de fim"),
        'numeroParticipantes': (ns.Evento, XSD.integer, "Número de participantes"),
        
        # Período
        'anoInicio': (ns.Periodo, XSD.gYear, "Ano de início"),
        'anoFim': (ns.Periodo, XSD.gYear, "Ano de fim"),
        'duracao': (ns.Periodo, XSD.integer, "Duração em anos"),
        
        # AreaConhecimento
        'nomeArea': (ns.AreaConhecimento, XSD.string, "Nome da área"),
        'descricao': (ns.Abstrato, XSD.string, "Descrição detalhada"),
        
        # Geral
        'identificador': (ns.Entidade, XSD.string, "Identificador único"),
        'notasAdicionais': (ns.Entidade, XSD.string, "Notas e observações"),
    }
    
    for prop_name, (domain, range_type, comment) in datatype_props.items():
        prop = ns[prop_name]
        g.add((prop, RDF.type, OWL.DatatypeProperty))
        g.add((prop, RDFS.label, Literal(prop_name, lang='pt')))
        g.add((prop, RDFS.domain, domain))
        g.add((prop, RDFS.range, range_type))
        g.add((prop, RDFS.comment, Literal(comment, lang='pt')))
    
    print(f"# {len(datatype_props)} datatype properties criadas")
    print ("# ====================================================================== ")     

    # ===================================================================
    # CARACTERÍSTICAS OWL AVANÇADAS
    # ===================================================================
    
    print("# Adicionando características OWL avançadas...")
    
    # Propriedades Simétricas
    simetricas = ['colaborouCom', 'casouCom', 'conheceu']
    for prop_name in simetricas:
        g.add((ns[prop_name], RDF.type, OWL.SymmetricProperty))
    print(f"# {len(simetricas)} propriedades simétricas")
    
    # Propriedades Transitivas
    transitivas = ['antes', 'depois', 'pertenceA']
    for prop_name in transitivas:
        g.add((ns[prop_name], RDF.type, OWL.TransitiveProperty))
    print(f"# {len(transitivas)} propriedades transitivas")
    
    # Propriedades Inversas
    inversas = [
        ('orientou', 'foiOrientadoPor'),
        ('inventou', 'construidaPor'),
        ('pai', 'filho'),
    ]
    for prop1, prop2 in inversas:
        g.add((ns[prop1], OWL.inverseOf, ns[prop2]))
    print(f"# {len(inversas)} pares de propriedades inversas")
    
    # Propriedades Funcionais
    funcionais = ['dataNascimento', 'dataMorte', 'pai', 'mae']
    for prop_name in funcionais:
        g.add((ns[prop_name], RDF.type, OWL.FunctionalProperty))
    print(f"# {len(funcionais)} propriedades funcionais")
    print ("# ====================================================================== ") 

    # ===================================================================
    # INSTÂNCIAS (Alan Turing e contexto)
    # ===================================================================
    
    print("# Criando instâncias...")
    
    # Alan Turing - instância múltipla (herança múltipla)
    g.add((ns.AlanTuring, RDF.type, ns.Matematico))
    g.add((ns.AlanTuring, RDF.type, ns.CientistaDaComputacao))
    g.add((ns.AlanTuring, RDF.type, ns.Criptografo))
    g.add((ns.AlanTuring, RDFS.label, Literal("Alan Mathison Turing", lang='pt')))
    g.add((ns.AlanTuring, ns.nomeCompleto, Literal("Alan Mathison Turing")))
    g.add((ns.AlanTuring, ns.primeiroNome, Literal("Alan")))
    g.add((ns.AlanTuring, ns.sobrenome, Literal("Turing")))
    g.add((ns.AlanTuring, ns.dataNascimento, Literal("1912-06-23", datatype=XSD.date)))
    g.add((ns.AlanTuring, ns.dataMorte, Literal("1954-06-07", datatype=XSD.date)))
    g.add((ns.AlanTuring, ns.nacionalidade, Literal("Britânico")))
    g.add((ns.AlanTuring, ns.profissao, Literal("Matemático, Criptanalista, Cientista da Computação")))
    g.add((ns.AlanTuring, ns.biografia, Literal("Pai da computação moderna e herói da Segunda Guerra Mundial")))
    
    # Lugares
    g.add((ns.Londres, RDF.type, ns.Cidade))
    g.add((ns.Londres, RDFS.label, Literal("Londres", lang='pt')))
    g.add((ns.Londres, ns.nomeLugar, Literal("Londres")))
    g.add((ns.Londres, ns.latitude, Literal("51.5074", datatype=XSD.decimal)))
    g.add((ns.Londres, ns.longitude, Literal("-0.1278", datatype=XSD.decimal)))
    
    g.add((ns.ReinoUnido, RDF.type, ns.Pais))
    g.add((ns.ReinoUnido, RDFS.label, Literal("Reino Unido", lang='pt')))
    g.add((ns.ReinoUnido, ns.nomeLugar, Literal("Reino Unido da Grã-Bretanha e Irlanda do Norte")))
    
    g.add((ns.Cambridge, RDF.type, ns.Cidade))
    g.add((ns.Cambridge, RDFS.label, Literal("Cambridge", lang='pt')))
    g.add((ns.Cambridge, ns.nomeLugar, Literal("Cambridge")))
    
    g.add((ns.BletchleyPark, RDF.type, ns.Edificio))
    g.add((ns.BletchleyPark, RDFS.label, Literal("Bletchley Park", lang='pt')))
    g.add((ns.BletchleyPark, ns.nomeLugar, Literal("Bletchley Park")))
    
    # Instituições
    g.add((ns.UniversidadeCambridge, RDF.type, ns.Universidade))
    g.add((ns.UniversidadeCambridge, RDFS.label, Literal("Universidade de Cambridge", lang='pt')))
    g.add((ns.UniversidadeCambridge, ns.nomeOrganizacao, Literal("University of Cambridge")))
    g.add((ns.UniversidadeCambridge, ns.dataFundacao, Literal("1209-01-01", datatype=XSD.date)))
    
    g.add((ns.UniversidadePrinceton, RDF.type, ns.Universidade))
    g.add((ns.UniversidadePrinceton, RDFS.label, Literal("Universidade de Princeton", lang='pt')))
    g.add((ns.UniversidadePrinceton, ns.nomeOrganizacao, Literal("Princeton University")))
    
    g.add((ns.GCCS, RDF.type, ns.OrganizacaoMilitar))
    g.add((ns.GCCS, RDFS.label, Literal("Government Code and Cypher School", lang='pt')))
    g.add((ns.GCCS, ns.nomeOrganizacao, Literal("Government Code and Cypher School")))
    
    # Máquinas
    g.add((ns.MaquinaTuring, RDF.type, ns.MaquinaComputacional))
    g.add((ns.MaquinaTuring, RDFS.label, Literal("Máquina de Turing", lang='pt')))
    g.add((ns.MaquinaTuring, ns.nomeMaquina, Literal("Máquina de Turing")))
    g.add((ns.MaquinaTuring, ns.anoConstrucao, Literal("1936", datatype=XSD.gYear)))
    
    g.add((ns.Bombe, RDF.type, ns.DispositivoCriptografico))
    g.add((ns.Bombe, RDFS.label, Literal("Bombe", lang='pt')))
    g.add((ns.Bombe, ns.nomeMaquina, Literal("Bombe Eletromecânica")))
    g.add((ns.Bombe, ns.peso, Literal("1000", datatype=XSD.decimal)))
    g.add((ns.Bombe, ns.altura, Literal("2.1", datatype=XSD.decimal)))
    g.add((ns.Bombe, ns.anoConstrucao, Literal("1939", datatype=XSD.gYear)))
    
    g.add((ns.Enigma, RDF.type, ns.DispositivoCriptografico))
    g.add((ns.Enigma, RDFS.label, Literal("Enigma", lang='pt')))
    g.add((ns.Enigma, ns.nomeMaquina, Literal("Máquina Enigma")))
    
    # Eventos
    g.add((ns.SegundaGuerraMundial, RDF.type, ns.EventoHistorico))
    g.add((ns.SegundaGuerraMundial, RDFS.label, Literal("Segunda Guerra Mundial", lang='pt')))
    g.add((ns.SegundaGuerraMundial, ns.nomeEvento, Literal("Segunda Guerra Mundial")))
    g.add((ns.SegundaGuerraMundial, ns.dataInicio, Literal("1939-09-01", datatype=XSD.date)))
    g.add((ns.SegundaGuerraMundial, ns.dataFim, Literal("1945-09-02", datatype=XSD.date)))
    
    # Documentos
    g.add((ns.ArtigoComputingMachinery, RDF.type, ns.Artigo))
    g.add((ns.ArtigoComputingMachinery, RDFS.label, Literal("Computing Machinery and Intelligence", lang='pt')))
    g.add((ns.ArtigoComputingMachinery, ns.titulo, Literal("Computing Machinery and Intelligence")))
    g.add((ns.ArtigoComputingMachinery, ns.dataPublicacao, Literal("1950-10-01", datatype=XSD.date)))
    g.add((ns.ArtigoComputingMachinery, ns.idioma, Literal("Inglês")))
    
    # Conceitos e Teorias
    g.add((ns.TesteTuring, RDF.type, ns.Conceito))
    g.add((ns.TesteTuring, RDFS.label, Literal("Teste de Turing", lang='pt')))
    g.add((ns.TesteTuring, ns.descricao, Literal("Teste para determinar se uma máquina pode exibir comportamento inteligente")))
    
    g.add((ns.TeoriaComputabilidade, RDF.type, ns.Teoria))
    g.add((ns.TeoriaComputabilidade, RDFS.label, Literal("Teoria da Computabilidade", lang='pt')))
    g.add((ns.TeoriaComputabilidade, ns.descricao, Literal("Teoria sobre o que pode ser computado")))
    
    # Áreas de Conhecimento
    g.add((ns.CienciaComputacao, RDF.type, ns.AreaConhecimento))
    g.add((ns.CienciaComputacao, RDFS.label, Literal("Ciência da Computação", lang='pt')))
    g.add((ns.CienciaComputacao, ns.nomeArea, Literal("Ciência da Computação")))
    
    g.add((ns.Criptografia, RDF.type, ns.AreaConhecimento))
    g.add((ns.Criptografia, RDFS.label, Literal("Criptografia", lang='pt')))
    g.add((ns.Criptografia, ns.nomeArea, Literal("Criptografia")))
    
    g.add((ns.InteligenciaArtificial, RDF.type, ns.AreaConhecimento))
    g.add((ns.InteligenciaArtificial, RDFS.label, Literal("Inteligência Artificial", lang='pt')))
    g.add((ns.InteligenciaArtificial, ns.nomeArea, Literal("Inteligência Artificial")))
    
    # Relações
    g.add((ns.AlanTuring, ns.nasceuEm, ns.Londres))
    g.add((ns.AlanTuring, ns.morreuEm, ns.ReinoUnido))
    g.add((ns.AlanTuring, ns.estudouEm, ns.UniversidadeCambridge))
    g.add((ns.AlanTuring, ns.doutoradoEm, ns.UniversidadePrinceton))
    g.add((ns.AlanTuring, ns.trabalhouEm, ns.GCCS))
    g.add((ns.AlanTuring, ns.trabalhouEm, ns.BletchleyPark))
    g.add((ns.AlanTuring, ns.inventou, ns.MaquinaTuring))
    g.add((ns.AlanTuring, ns.inventou, ns.Bombe))
    g.add((ns.AlanTuring, ns.desenvolveu, ns.TeoriaComputabilidade))
    g.add((ns.AlanTuring, ns.propôs, ns.TesteTuring))
    g.add((ns.AlanTuring, ns.publicou, ns.ArtigoComputingMachinery))
    g.add((ns.AlanTuring, ns.especializadoEm, ns.CienciaComputacao))
    g.add((ns.AlanTuring, ns.especializadoEm, ns.Criptografia))
    g.add((ns.AlanTuring, ns.contribuiuPara, ns.InteligenciaArtificial))
    g.add((ns.AlanTuring, ns.participouDe, ns.SegundaGuerraMundial))
    
    g.add((ns.BletchleyPark, ns.localizadaEm, ns.ReinoUnido))
    g.add((ns.UniversidadeCambridge, ns.localizadaEm, ns.Cambridge))
    g.add((ns.Cambridge, ns.localizadaEm, ns.ReinoUnido))
    g.add((ns.Londres, ns.localizadaEm, ns.ReinoUnido))
    
    g.add((ns.Bombe, ns.usadaEm, ns.BletchleyPark))
    g.add((ns.MaquinaTuring, ns.baseadaEm, ns.TeoriaComputabilidade))
    
    g.add((ns.SegundaGuerraMundial, ns.ocorreuEm, ns.ReinoUnido))
    g.add((ns.GCCS, ns.pertenceA, ns.ReinoUnido))
    
    print(f"# 20+ instâncias criadas com relações completas")
    print ("# ====================================================================== \n") 
    
    # ===================================================================
    # SALVAR
    # ===================================================================
    
    arquivo_saida = 'ontologia_otimizada.ttl'
    DIR_DATASET = "C:\\Users\\alexa\\Atividades\\LocalDeepSeek\\Dataset\\"

    g.serialize(DIR_DATASET + arquivo_saida, format='turtle')   
    return arquivo_saida

```

```python

class OntologyQualityAnalyzer:
    """Analisador de qualidade de ontologias"""
    
    def __init__(self, ontology_file):
        """Inicializa o analisador"""
        self.ontology_file = ontology_file
        self.graph = Graph()
        self.metrics = {}
        self.issues = []
        self.recommendations = []
        
        # Carregar ontologia
        self._load_ontology()
        
    def _load_ontology(self):
        """Carrega a ontologia"""
        self.graph.parse(self.ontology_file, format='turtle')
        
    def analyze_all(self):
        """Executa todas as análises"""
        print ("# ====================================================================== ")
        print ("# Iniciando aInálise de qualidade")
        print ("# ====================================================================== \n")        
        
        # Análises básicas
        self._analyze_structure()
        
        # Análises de qualidade
        self._analyze_completeness()
        self._analyze_consistency()
        self._analyze_conciseness()
        self._analyze_semantic_richness()
        self._analyze_connectivity()
        self._analyze_hierarchy()
        self._analyze_coverage()
        self._analyze_documentation()
        
        # Calcular score geral
        self._calculate_overall_score()
        
        
    def _analyze_structure(self):
        """Analisa estrutura básica"""
        print ("# ====================================================================== ")
        print ("# ESTRUTURA BÁSICA")
        print ("# ====================================================================== ")
        
        # Contar elementos
        classes = list(self.graph.subjects(RDF.type, OWL.Class))
        obj_props = list(self.graph.subjects(RDF.type, OWL.ObjectProperty))
        data_props = list(self.graph.subjects(RDF.type, OWL.DatatypeProperty))
        
        # Contar instâncias
        instancias = set()
        for cls in classes:
            for inst in self.graph.subjects(RDF.type, cls):
                if inst not in classes:
                    instancias.add(inst)
        
        self.metrics['total_classes'] = len(classes)
        self.metrics['total_object_properties'] = len(obj_props)
        self.metrics['total_datatype_properties'] = len(data_props)
        self.metrics['total_instancias'] = len(instancias)
        self.metrics['total_triplas'] = len(self.graph)
        
        self.metrics['classes'] = classes
        self.metrics['object_properties'] = obj_props
        self.metrics['datatype_properties'] = data_props
        self.metrics['instancias'] = list(instancias)
        
        print(f"# Triplas totais ......: {self.metrics['total_triplas']}")
        print(f"# Classes .............: {self.metrics['total_classes']}")
        print(f"# Object Properties....: {self.metrics['total_object_properties']}")
        print(f"# Datatype Properties..: {self.metrics['total_datatype_properties']}")
        print(f"# Instâncias ..........: {self.metrics['total_instancias']}")
        
    def _analyze_completeness(self):
        """Analisa completude"""
        print ("# ====================================================================== ")
        print ("# COMPLETUDE (Completeness)")
        print ("# ====================================================================== ")        
        
        classes = self.metrics['classes']
        props = self.metrics['object_properties'] + self.metrics['datatype_properties']
        instancias = self.metrics['instancias']
        
        # Classes com labels
        classes_com_label = sum(1 for cls in classes 
                               if list(self.graph.objects(cls, RDFS.label)))
        
        # Classes com comentários
        classes_com_comentario = sum(1 for cls in classes 
                                    if list(self.graph.objects(cls, RDFS.comment)))
        
        # Propriedades com domínio
        props_com_dominio = sum(1 for prop in props 
                               if list(self.graph.objects(prop, RDFS.domain)))
        
        # Propriedades com range
        props_com_range = sum(1 for prop in props 
                             if list(self.graph.objects(prop, RDFS.range)))
        
        # Instâncias com labels
        inst_com_label = sum(1 for inst in instancias 
                            if list(self.graph.objects(inst, RDFS.label)))
        
        # Calcular score
        scores = [
            classes_com_label / len(classes) if classes else 1,
            classes_com_comentario / len(classes) if classes else 1,
            props_com_dominio / len(props) if props else 1,
            props_com_range / len(props) if props else 1,
            inst_com_label / len(instancias) if instancias else 1
        ]
        
        score_completude = sum(scores) / len(scores) * 100      
        self.metrics['completeness_score'] = round(score_completude, 1)
        
        print(f"# Classes com labels ........: {classes_com_label}/{len(classes)} ({classes_com_label/len(classes)*100:.1f}%)")
        print(f"# Classes com comentários....: {classes_com_comentario}/{len(classes)} ({classes_com_comentario/len(classes)*100:.1f}%)")
        print(f"# Propriedades com domínio...: {props_com_dominio}/{len(props)} ({props_com_dominio/len(props)*100:.1f}%)")
        print(f"# Propriedades com range.....: {props_com_range}/{len(props)} ({props_com_range/len(props)*100:.1f}%)")
        print(f"# Instâncias com labels......: {inst_com_label}/{len(instancias)} ({inst_com_label/len(instancias)*100 if instancias else 0:.1f}%)")
        print(f"# SCORE DE COMPLETUDE .......: {score_completude:.1f}%")
        
    def _analyze_consistency(self):
        """Analisa consistência"""
        print ("# ====================================================================== ")
        print ("# CONSISTÊNCIA (Consistency)")
        print ("# ====================================================================== ")     
        
        classes = self.metrics['classes']
        props = self.metrics['object_properties'] + self.metrics['datatype_properties']
        
        # Classes órfãs
        orfas = sum(1 for cls in classes
                   if not list(self.graph.objects(cls, RDFS.subClassOf)) and
                      not list(self.graph.subjects(RDFS.subClassOf, cls)))
        
        # Propriedades sem domínio nem range
        props_sem_dom_range = sum(1 for prop in props
                                 if not list(self.graph.objects(prop, RDFS.domain)) and
                                    not list(self.graph.objects(prop, RDFS.range)))
        
        # Ciclos (simplificado)
        ciclos = 0
        
        # Classes com nomes inconsistentes
        nomes_inconsistentes = 0
        
        problemas = orfas + props_sem_dom_range + ciclos + nomes_inconsistentes
        score = 100 - (problemas * 10) if problemas < 10 else 0
        
        self.metrics['consistency_score'] = max(0, score)
        
        print(f"# Classes órfãs (sem hierarquia) .......: {orfas}")
        print(f"# Propriedades sem domínio nem range ...: {props_sem_dom_range}")
        print(f"# Ciclos na hierarquia .................: {ciclos}")
        print(f"# Classes com nome inconsistente .......: {nomes_inconsistentes}")
        print(f"# SCORE DE CONSISTÊNCIA ................: {score:.1f}%")
        
        if orfas > 0:
            self.issues.append({
                'severidade': 'média',
                'descricao': f"Classes órfãs sem hierarquia: {orfas}"
            })
            
    def _analyze_conciseness(self):
        """Analisa concisão"""
        print ("# ====================================================================== ")
        print ("# CONCISÃO (Conciseness)")
        print ("# ====================================================================== ")          
        
        classes = self.metrics['classes']
        props = self.metrics['object_properties'] + self.metrics['datatype_properties']
        
        # Labels duplicados
        labels = [str(label) for cls in classes 
                 for label in self.graph.objects(cls, RDFS.label)]
        contador = Counter(labels)
        labels_dup = sum(1 for count in contador.values() if count > 1)
        
        # Nomes similares
        from difflib import SequenceMatcher
        nomes = [str(prop).split('#')[-1].split('/')[-1] for prop in props]
        similares = 0
        for i, nome1 in enumerate(nomes):
            for nome2 in nomes[i+1:]:
                if SequenceMatcher(None, nome1.lower(), nome2.lower()).ratio() > 0.8:
                    similares += 1
        
        # Triplas redundantes
        redundantes = 0
        
        problemas = labels_dup + (similares // 2) + redundantes
        score = 100 - (problemas * 2)
        
        self.metrics['conciseness_score'] = max(0, min(100, score))
        
        print(f"# Classes com labels duplicados ............: {labels_dup}")
        print(f"# Propriedades com nomes muito similares ...: {similares}")
        print(f"# Triplas redundantes ......................: {redundantes}")
        print(f"# SCORE DE CONCISÃO ........................: {score:.1f}%")
        
    def _analyze_semantic_richness(self):
        """Analisa riqueza semântica"""
        print ("# ====================================================================== ")
        print ("# RIQUEZA SEMÂNTICA (Semantic Richness)")
        print ("# ====================================================================== ")  
        
        classes = self.metrics['classes']
        obj_props = self.metrics['object_properties']
        data_props = self.metrics['datatype_properties']
        
        num_classes = len(classes)
        
        relacoes_por_classe = len(obj_props) / num_classes if num_classes > 0 else 0
        atributos_por_classe = len(data_props) / num_classes if num_classes > 0 else 0
        
        # Profundidade da hierarquia
        def calcular_profundidade(cls, visitados=None):
            if visitados is None:
                visitados = set()
            if cls in visitados:
                return 0
            visitados.add(cls)
            superclasses = list(self.graph.objects(cls, RDFS.subClassOf))
            if not superclasses:
                return 1
            return 1 + max(calcular_profundidade(sup, visitados) for sup in superclasses)
        
        profundidades = [calcular_profundidade(cls) for cls in classes]
        profundidade_max = max(profundidades) if profundidades else 0
        
        # Características OWL
        caracteristicas = 0
        if list(self.graph.subjects(RDF.type, OWL.SymmetricProperty)):
            caracteristicas += 1
        if list(self.graph.subjects(RDF.type, OWL.TransitiveProperty)):
            caracteristicas += 1
        if list(self.graph.subjects(RDF.type, OWL.FunctionalProperty)):
            caracteristicas += 1
        if list(self.graph.subjects(RDF.type, OWL.InverseFunctionalProperty)):
            caracteristicas += 1
        
        # Score
        score_rel = min(100, relacoes_por_classe * 50)
        score_atr = min(100, atributos_por_classe * 50)
        score_hier = min(100, profundidade_max * 20)
        score_owl = (caracteristicas / 4) * 100
        
        score = (score_rel + score_atr + score_hier + score_owl) / 4        
        self.metrics['semantic_richness_score'] = round(score, 1)
        
        print(f"# Relações por classe ................: {relacoes_por_classe:.2f}")
        print(f"# Atributos por classe................: {atributos_por_classe:.2f}")
        print(f"# Profundidade máxima da hierarquia...: {profundidade_max}")
        print(f"# Características OWL avançadas usadas: {caracteristicas}/4")
        print(f"# SCORE DE RIQUEZA SEMÂNTICA..........: {score:.1f}%")
        
        if profundidade_max < 3:
            self.issues.append({
                'severidade': 'baixa',
                'descricao': "Hierarquia rasa (profundidade < 3)"
            })
            
    def _analyze_connectivity(self):
        """Analisa conectividade"""
        print ("# ====================================================================== ")
        print ("# CONECTIVIDADE (Connectivity)")
        print ("# ====================================================================== ")  
        
        classes = self.metrics['classes']
        
        # Construir grafo NetworkX
        G = nx.DiGraph()
        for cls in classes:
            G.add_node(str(cls))
            for superclasse in self.graph.objects(cls, RDFS.subClassOf):
                if superclasse in classes:
                    G.add_edge(str(cls), str(superclasse))
        
        G_undir = G.to_undirected()
        componentes = list(nx.connected_components(G_undir))
        num_componentes = len(componentes)
        
        maior_componente = max(len(comp) for comp in componentes) if componentes else 0
        porc_maior = (maior_componente / len(classes) * 100) if classes else 0
        
        nos_isolados = sum(1 for comp in componentes if len(comp) == 1)
        densidade = nx.density(G_undir) if len(G_undir) > 1 else 0
        
        # Score
        score_comp = 100 - (num_componentes - 1) * 20
        score_isol = 100 - nos_isolados * 10
        score_dens = densidade * 100
        
        score = (score_comp * 0.4 + score_isol * 0.3 + score_dens * 0.3)
        
        self.metrics['connectivity_score'] = max(0, min(100, round(score, 1)))
        
        print(f"# Componentes conectados.....: {num_componentes}")
        print(f"# Maior componente...........: {maior_componente} nós ({porc_maior:.1f}%)")
        print(f"# Nós isolados...............: {nos_isolados}")
        print(f"# Densidade do grafo.........: {densidade:.4f}")
        print(f"# SCORE DE CONECTIVIDADE.....: {score:.1f}%")
        
        if num_componentes > 1:
            self.issues.append({
                'severidade': 'média',
                'descricao': f"Grafo fragmentado em {num_componentes} componentes"
            })
            
    def _analyze_hierarchy(self):
        """Analisa hierarquia"""
        print ("# ====================================================================== ")
        print ("# HIERARQUIA (Hierarchy Structure)")
        print ("# ====================================================================== ")  
        
        classes = self.metrics['classes']
        
        # Profundidade
        def calcular_profundidade(cls, visitados=None):
            if visitados is None:
                visitados = set()
            if cls in visitados:
                return 0
            visitados.add(cls)
            superclasses = list(self.graph.objects(cls, RDFS.subClassOf))
            if not superclasses:
                return 1
            return 1 + max(calcular_profundidade(sup, visitados) for sup in superclasses)
        
        profundidades = [calcular_profundidade(cls) for cls in classes]
        prof_max = max(profundidades) if profundidades else 0
        prof_media = sum(profundidades) / len(profundidades) if profundidades else 0
        
        min_prof = min(profundidades) if profundidades else 0
        balanceamento = min_prof / prof_max if prof_max > 0 else 1
        
        # Score
        score_prof = min(100, prof_max * 20)
        score_bal = balanceamento * 100
        score = (score_prof + score_bal) / 2
        
        self.metrics['hierarchy_score'] = round(score, 1)
        
        print(f"# Profundidade máxima........: {prof_max}")
        print(f"# Profundidade média.........: {prof_media:.2f}")
        print(f"# Balanceamento..............: {balanceamento:.2f}")
        print(f"# SCORE DE HIERARQUIA........: {score:.1f}%")
        
    def _analyze_coverage(self):
        """Analisa cobertura"""
        print ("# ====================================================================== ")
        print ("# COBERTURA (Schema Coverage)")
        print ("# ====================================================================== ")  
        
        classes = self.metrics['classes']
        
        # Classes com instâncias
        classes_com_inst = sum(1 for cls in classes 
                              if list(self.graph.subjects(RDF.type, cls)))
        
        total_classes = len(classes)
        porc_cobertura = (classes_com_inst / total_classes * 100) if total_classes > 0 else 0
        
        # Instâncias por classe
        instancias_por_classe = []
        for cls in classes:
            num_inst = len(list(self.graph.subjects(RDF.type, cls)))
            if num_inst > 0:
                instancias_por_classe.append(num_inst)
        
        media_inst = sum(instancias_por_classe) / len(instancias_por_classe) if instancias_por_classe else 0
        mediana_inst = sorted(instancias_por_classe)[len(instancias_por_classe)//2] if instancias_por_classe else 0
        
        self.metrics['coverage_score'] = round(porc_cobertura, 1)
        
        print(f"# Classes com instâncias...........: {classes_com_inst}/{total_classes} ({porc_cobertura:.1f}%)")
        print(f"# Média de instâncias por classe...: {media_inst:.1f}")
        print(f"# Mediana de instâncias por classe.: {mediana_inst}")
        print(f"# SCORE DE COBERTURA...............: {porc_cobertura:.1f}%")
        
        if porc_cobertura < 50:
            self.issues.append({
                'severidade': 'média',
                'descricao': f"Apenas {porc_cobertura:.0f}% das classes têm instâncias"
            })
            self.recommendations.append("Adicionar instâncias para classes que não têm exemplos")
            
    def _analyze_documentation(self):
        """Analisa documentação"""
        print ("# ====================================================================== ")
        print ("# DOCUMENTAÇÃO (Documentation Quality)")
        print ("# ====================================================================== ") 
        
        classes = self.metrics['classes']
        props = self.metrics['object_properties'] + self.metrics['datatype_properties']
        todos = classes + props
        
        # Labels em português
        labels_pt = sum(1 for elem in todos
                       for label in self.graph.objects(elem, RDFS.label)
                       if label.language == 'pt')
        
        total_labels = sum(1 for elem in todos 
                          if list(self.graph.objects(elem, RDFS.label)))
        
        # Comentários informativos
        comentarios_info = sum(1 for elem in todos
                              for comment in self.graph.objects(elem, RDFS.comment)
                              if len(str(comment)) >= 20)
        
        score_idioma = (labels_pt / total_labels * 100) if total_labels > 0 else 0
        score_com = (comentarios_info / len(todos) * 100) if todos else 0        
        score = (score_idioma + score_com) / 2        
        self.metrics['documentation_score'] = round(score, 1)
        
        print(f"# Labels em português...........: {labels_pt}/{total_labels} ({score_idioma:.1f}%)")
        print(f"# Comentários informativos......: {comentarios_info}/{len(todos)} ({score_com:.1f}%)")
        print(f"# SCORE DE DOCUMENTAÇÃO.........: {score:.1f}%")
                
    def _calculate_overall_score(self):
        """Calcula score geral"""
        scores = [
            self.metrics.get('completeness_score', 0),
            self.metrics.get('consistency_score', 0),
            self.metrics.get('conciseness_score', 0),
            self.metrics.get('semantic_richness_score', 0),
            self.metrics.get('connectivity_score', 0),
            self.metrics.get('hierarchy_score', 0),
            self.metrics.get('coverage_score', 0),
            self.metrics.get('documentation_score', 0)
        ]
        
        self.metrics['overall_score'] = round(sum(scores) / len(scores), 1)
        
        dimensoes = [
            ('completeness_score', 'Completude'),
            ('consistency_score', 'Consistência'),
            ('conciseness_score', 'Concisão'),
            ('semantic_richness_score', 'Riqueza Semântica'),
            ('connectivity_score', 'Conectividade'),
            ('hierarchy_score', 'Qualidade da Hierarquia'),
            ('coverage_score', 'Cobertura do Schema'),
            ('documentation_score', 'Qualidade da Documentação')
        ]
        
        for chave, nome in dimensoes:
            score_dim = self.metrics.get(chave, 0)
            barra = '█' * int(score_dim / 5)
            print(f"# {nome:.<30} {score_dim:>5.1f}% {barra}")
        print ("# ====================================================================== ") 
            
        # Problemas
        if self.issues:
            print("# ====================================================================== ") 
            print(f"# PROBLEMAS IDENTIFICADOS ({len(self.issues)}):")
            print ("# ====================================================================== ") 
            
            por_severidade = {'crítica': [], 'alta': [], 'média': [], 'baixa': []}
            for issue in self.issues:
                por_severidade[issue['severidade']].append(issue['descricao'])
            
            for sev in ['crítica', 'alta', 'média', 'baixa']:
                if por_severidade[sev]:
                    print(f"# {sev.upper()}:")
                    for desc in por_severidade[sev]:
                        print(f"# {desc}")
                
        # Recomendações
        if self.recommendations:
            print("# ====================================================================== ") 
            print("# RECOMENDAÇÕES DE MELHORIA")
            print("# ====================================================================== ") 
            for i, rec in enumerate(self.recommendations, 1):
                print(f"# {i}. {rec}")
        
        print("# ====================================================================== ") 
        
    def export_json(self, output_file='quality_metrics.json'):
        """Exporta métricas para JSON"""
        dados = {
            'arquivo': self.ontology_file,
            'data_analise': datetime.now().isoformat(),
            'score_geral': self.metrics.get('overall_score', 0),
            'scores': {
                'completude': self.metrics.get('completeness_score', 0),
                'consistencia': self.metrics.get('consistency_score', 0),
                'concisao': self.metrics.get('conciseness_score', 0),
                'riqueza': self.metrics.get('semantic_richness_score', 0),
                'conectividade': self.metrics.get('connectivity_score', 0),
                'hierarquia': self.metrics.get('hierarchy_score', 0),
                'cobertura': self.metrics.get('coverage_score', 0),
                'documentacao': self.metrics.get('documentation_score', 0)
            },
            'problemas': self.issues,
            'recomendacoes': self.recommendations
        }
        
        DIR_OUTPUT = "C:\\Users\\alexa\\Atividades\\LocalDeepSeek\\Output\\"
        with open(DIR_OUTPUT + output_file, 'w', encoding='utf-8') as f:
            json.dump(dados, f, indent=2, ensure_ascii=False)
        
        print(f"# Métricas exportadas: {output_file}")
```
